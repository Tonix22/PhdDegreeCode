```python
# Jupyter Notebook: Summary and Explanation of "Incoherent DPSK Diversity Reception in Doubly Selective Channels Based on Virtual Trajectories"

# Import necessary libraries
import sympy as sp
from sympy import symbols, summation, conjugate, IndexedBase, Eq

# Introduction
"""
This notebook summarizes the paper "Incoherent DPSK Diversity Reception in Doubly Selective Channels Based on Virtual Trajectories" by Pena-Campos et al., using symbolic equations with SymPy for clarity.

# Key Concepts
- DPSK (Differential Phase Shift Keying): A modulation scheme that encodes information based on phase differences.
- Doubly Selective Channels (DSC): Wireless channels with variations in both frequency and time, caused by multipath propagation and Doppler shifts.
- Virtual Trajectories (VT): Technique used to simplify the detection process by representing the channel effects as virtual antennas.

# Signal Model
The received signal model for DSC:
"""

# Define symbolic variables
n, l, N, L = symbols('n l N L', integer=True)
x = IndexedBase('x')
h = IndexedBase('h')
w = IndexedBase('w')

# Received signal equation (1) in the paper
y_n = summation(h[n, l] * x[(n - l) % N], (l, 0, L-1)) + w[n]
display(Eq(symbols('y[n]'), y_n))

"""
Where:
- x[n]: Transmitted signal.
- h[n, l]: Channel response at time n and delay l.
- w[n]: White noise at receiver.
"""

# Basis Expansion Model (BEM) for channel representation (Eq. 2 in the paper)
MD, Mτ, q, r = symbols('M_D M_tau q r', integer=True)
α = IndexedBase('alpha')
φ_I = IndexedBase('phi_I')
φ_II = IndexedBase('phi_II')

h_nl = summation(summation(α[q, r]*φ_I[q, n]*φ_II[r, l], (q, 0, MD-1)), (r, 0, Mτ-1))
display(Eq(h[n, l], h_nl))

# DPSK Modulation
"""
Information bits are mapped to DPSK symbols:
- Differential encoding:
"""

β = IndexedBase('beta')
d = IndexedBase('d')
i = symbols('i', integer=True)

β_i = β[i-1]*d[i]
display(Eq(β[i], β_i))

"""
Where:
- d[i]: Information symbol.
- β[i]: Differentially encoded symbol.
"""

# Receiver Model and Virtual Trajectories
"""
Received signal vector using Virtual Trajectories representation:
"""
Ψ, Ω, ϑ = symbols('Psi Omega vartheta')
y = symbols('y')
w = symbols('w')

receiver_eq = Eq(y, Ψ * Ω * β + w)
display(receiver_eq)

"""
Where:
- Ψ: Matrix formed by basis functions.
- Ω: Channel frequency response.
- β: Differentially encoded transmitted symbols.
"""

# Autocorrelation-based Detection (Simplified)
"""
Detection using autocorrelation, avoiding explicit channel estimation:
"""
MD = symbols('M_D', integer=True)

ϑ_hat = IndexedBase('vartheta_hat')
q = symbols('q', integer=True)

# Decision rule
AC_decision_rule = summation(ϑ_hat[q, i]*conjugate(ϑ_hat[q, i-1]), (q, 0, MD-1))
detected_symbol = sp.Function('argmax')(sp.re(AC_decision_rule))
display(Eq(symbols('d_hat[i]'), detected_symbol))

"""
- ϑ̂[q,i]: Estimated virtual trajectories.
- d̂[i]: Detected DPSK symbol.

# Main advantages of this method:
- Lower computational complexity (O(N log2 N)).
- Avoids explicit channel estimation.
- Robustness against Doppler-induced intercarrier interference (ICI).

This summary highlights essential aspects using symbolic representation, aiding understanding of the paper's contributions.
"""
```
